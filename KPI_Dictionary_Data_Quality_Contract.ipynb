```python
import numpy as np
import pandas as pd
```

## **Dataset** **Loading**

```python
df=pd.read_csv('retail-orders-raw.csv')
print(df)
print(df.head())
print('Rows',df.shape[0])
print('column',df.shape[1])
print(df.info())
```

## **DataSet** **Overview**

```python
print(df.columns.tolist())
```

```python
print(df.info())
```

```python
print(df.describe(include='all'))
```

## **Data Quality- Completeness**

```python
# Missing Values -In Completeness Check
missing_count = df.isnull().sum()
print(missing_count)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage.round(2))
```

## **Data Quality - Uniqueness**

```python
# Uniqueness Check

duplicate_count = df["order_id"].duplicated().sum()

print("Duplicate Order IDs:", duplicate_count)
```

```python
df[df["order_id"].duplicated(keep=False)]
```

```python
df["quantity"] = pd.to_numeric(df["quantity"], errors="coerce")

df["unit_price"] = pd.to_numeric(df["unit_price"], errors="coerce")

df["discount_pct"] = pd.to_numeric(df["discount_pct"], errors="coerce")

print(df.dtypes)
```

```python
print("Invalid/Converted Quantity:", df["quantity"].isna().sum())
print("Invalid/Converted Unit Price:", df["unit_price"].isna().sum())
print("Invalid/Converted Discount:", df["discount_pct"].isna().sum())
print(df.dtypes)
```

## **Data Quality - Validity**

```python
# Data Validity Check
df["order_date"] = pd.to_datetime(
    df["order_date"],
    errors="coerce",
    dayfirst=True
)

df["order_date"]
```

```python
# Check Invalid dates
invalid_dates = df["order_date"].isna().sum()

print("Invalid dates:", invalid_dates)
```

```python
# Quantity Validity Check

invalid_quantity = (df["quantity"] <= 0).sum()

print("Invalid Quantity:", invalid_quantity)
df[df["quantity"] <= 0]
```

```python
# Unit Price Validity Check
invalid_Unit_price = (df["unit_price"] <= 0).sum()

print("Invalid Unit_Price:", invalid_Unit_price)
```

```python
# Discount Validity Check

invalid_discount = (
    (df["discount_pct"] < 0) |
    (df["discount_pct"] > 100)
).sum()

print("Invalid Discount:", invalid_discount)

```

```python
# Payment Status check Validity
df["payment_status"] = df["payment_status"].str.strip().str.title()

print(df["payment_status"].unique())


#Valid check
valid_status = ["Paid", "Pending", "Failed"]

invalid_payment = (~df["payment_status"].isin(valid_status)).sum()

print("Invalid Payment Status:", invalid_payment)
```

```python
# Customer Segment Consistency

df['customer_segment']=df["customer_segment"].str.strip().str.title()

print(df["customer_segment"].unique())
```

```python
# City Check
df["city"] = df["city"].str.strip().str.title()
print(df["city"].unique())
```

## **Data Quality - Consistency**

```python
#  Consistency Check — Revenue

df["gross_amount"] = df["quantity"] * df["unit_price"]
print(df["gross_amount"].unique())
```

```python
print("Missing Gross Amount:", df["gross_amount"].isna().sum())
```

```python
inconsistent_revenue = df[
    df["gross_amount"].isna()
]

print("Inconsistent/Missing Revenue Records:", len(inconsistent_revenue))
```

```python
df["discount_amount"] = (
    df["gross_amount"] * df["discount_pct"].fillna(0) / 100
)

df["net_sales"] = df["gross_amount"] - df["discount_amount"]
df[
    ["order_id", "quantity", "unit_price",
     "discount_pct", "gross_amount",
     "discount_amount", "net_sales"]
]
```

Data Quality - Freshness

```python
# Fresh Check
latest_date = df["order_date"].max()

print("Latest Order Date:", latest_date)

oldest_date = df["order_date"].min()

print("Oldest Order Date:", oldest_date)
```

## **KPI Calulations**

```python
# Calculate KPI
# Total revenue
total_revenue = df["net_sales"].sum()

print("Total Revenue:", total_revenue)

# Total Order
# Duplicate not allowed
total_orders = df["order_id"].nunique()

print("Total Orders:", total_orders)

# Total Unit
total_units = df["quantity"].sum()

print("Total Units Sold:", total_units)

# Average Order Value

average_order_value = total_revenue / total_orders

print("Average Order Value:", round(average_order_value, 2))

# Paid Orders
paid_orders = df.loc[
    df["payment_status"] == "Paid",
    "order_id"
].nunique()

print("Paid Orders:", paid_orders)

# Pending Orders
pending_orders = df.loc[
    df["payment_status"] == "Pending",
    "order_id"
].nunique()

print("Pending Orders:", pending_orders)

# Average Discount
average_discount = df["discount_pct"].fillna(0).mean()

print("Average Discount %:", round(average_discount, 2))


# Revenue by Category
revenue_by_category = (
    df.groupby("category")["net_sales"]
    .sum()
    .sort_values(ascending=False)
)

print(revenue_by_category)
```

## **KPI Summary Table**

```python
# KPI Summary Table
kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Revenue",
        "Total Orders",
        "Total Units Sold",
        "Average Order Value",
        "Paid Orders",
        "Pending Orders",
        "Average Discount %",
        "Number of Cities"
    ],
    "Value": [
        total_revenue,
        total_orders,
        total_units,
        average_order_value,
        paid_orders,
        pending_orders,
        average_discount,
        df["city"].nunique()
    ]
})

kpi_summary
```

# **Final** **Data** **Quality** **Report**



```python
quality_report = pd.DataFrame({
    "Quality_Dimension": [
        "Completeness",
        "Uniqueness",
        "Validity - Date",
        "Validity - Quantity",
        "Validity - Unit Price",
        "Validity - Discount",
        "Validity - Payment Status",
        "Consistency - Customer Segment",
        "Freshness"
    ],
    "Result": [
        f"{missing_percentage.mean():.2f}% average missing",
        f"{duplicate_count} duplicate Order IDs",
        f"{invalid_dates} invalid dates",
        f"{invalid_quantity} invalid quantities",
        f"{invalid_Unit_price} invalid_Unit_prices",
        f"{invalid_discount} invalid discounts",
        f"{invalid_payment} invalid payment statuses",
        "Standardized",
        str(latest_date.date())
    ]
})

quality_report
```

```python
kpi_summary.to_csv("KPI_Summary.csv", index=False)
```

```python
with pd.ExcelWriter("KPI_Dictionary_Data_Quality_Report.xlsx") as writer:
    kpi_summary.to_excel(writer, sheet_name="KPI Summary", index=False)
    quality_report.to_excel(writer, sheet_name="Quality Report", index=False)
```
